Bar chart
--
is a graph that presents data with rectangular bars.  

They can be drawn either vertically or horizontally

One can draw using **Matplotlib**

**bar(x, height, width, bottom, align)**

It also has other options such as color, linewidth, ...


````python
import numpy as np
import matplotlib.pyplot as plot


# creating the dataset
data = {'R':200, 'C++':115, 'Ada':330,
		'Python':135}
courses = list(data.keys())
values = list(data.values())

fig = plot.figure(figsize = (10, 5))

# creating the bar plot
plot.bar(courses, values, color ='teal',
		width = 0.3)

plot.xlabel("Course")
plot.ylabel("No. of students ")
plot.title("Students enrolled")
plot.show()
````

````python
# creating the bar plot
plot.barh(courses, values, color ='teal')

plot.ylabel("Course")
plot.xlabel("No. of students ")
plot.title("Students enrolled")
plot.show()
````

Plotting histogram
--
Histogram is a graphical representation, which is quite similar to a bar graph.

They plots the frequency distribution of numeric aray by splittung it to small equal-sized bins.
* they help in understanding the probabilistic distribution of the numerical data
* one can find out the type of distribution (*Gaussian*, *skewed*, *exponential*, *step*, ...)

* Histograms help us to see possible outliers.


plt.hist(data, bins= number of bins)

other parameters are

* **range** specifies lower and upper range of the bins
* **density** Boolean, Normalization to unity
* **histtype** type of histogram  "bar", "barstacked", "step"  , "stepfilled".

Box plot or Whisker plot
--

displays summary of a set of data containing the minimum, first quartile, median, third quartile, and maximum.

*Quartile* divied the number of data points into four parts (*Qaurter*).
*  First quartile (Q1) is defined as the middle number between the smallest number (minimum) and the median of the data set. It is also known as the lower quartile, as 25% of the data is below this point.
* Second quartile (Q2) is the median of a data set; thus 50% of the data lies below this point.
* Third quartile (Q3) is the middle value between the median and the highest value (maximum) of the data set. It is known as the upper quartile, as 75% of the data lies below this point.

* *meanline*  having boolean value try to render meanline as full width of box

````python
import numpy as np
import matplotlib.pyplot as plt

# Create a grid of points
x = np.linspace(-5, 5, 25)
y = np.linspace(-5, 5, 25)
X, Y = np.meshgrid(x, y)

# Tornado-like wind field:
# Swirling motion + inward pull toward center
r2 = X**2 + Y**2 + 0.1  # radius^2 (avoid divide-by-zero)
U = -Y / r2 - 0.1*X / np.sqrt(r2)   # horizontal component (swirl + inward)
V = X / r2 - 0.1*Y / np.sqrt(r2)    # vertical component (swirl + inward)

# Speed magnitude
speed = np.sqrt(U**2 + V**2)

# Plot quiver
plt.figure(figsize=(8,8))
q = plt.quiver(X, Y, U, V, speed, cmap='plasma', scale=20, width=0.005)
plt.colorbar(q, label="Wind speed (arbitrary units)")

# Tornado core
plt.plot(0, 0, 'r^', markersize=16, label="Tornado core")

plt.title("Simulated Tornado Wind Field (Quiver Plot)")
plt.xlabel("X position (km)")
plt.ylabel("Y position (km)")
plt.legend()
plt.axis('equal')
plt.show()
````

Correlation matrix plot
--

Simplest defintion is:
A correlation coefficient is a numerical measure of some type of correlation, meaning a statistical relationship between two variables
* It indicates the strength of relationship between two variables.

 Mainly Pearson's coefficint is used.
* it is estimated by dividing the covariance between two variables by product of standard deviation of two variables.

 $ \rho(x,y) = \frac{Cov(x,y)}{\sigma_x \sigma_y}$

 $ = \frac{ N  (\Sigma xy) - \Sigma x \Sigma y}{\sqrt{ [ N\Sigma x^2 - (\Sigma x)^2 ] ~~[ N \Sigma y^2 - (\Sigma y)^2] } }$


The value of $\rho$ varies from -1 to +1.
If $\rho =1$ indidcate strong relation between $x$ and $y$

Value of 0 means $x$ and $y$ are independent



````python
from matplotlib import pyplot
import numpy as np
import matplotlib
from random import seed
from random import randint
np.random.seed(1000)
x= np.random.randint(0,100,1000)
y= x+ np.random.randint(0,20,1000)
print(np.corrcoef(x,y))
matplotlib.pyplot.scatter(x,y)
matplotlib.pyplot.show()
````

````python
import json
import folium
import pandas as pd
from folium.features import GeoJsonTooltip

GEOJSON_PATH="All_India_pincode_Boundary-19312.geojson"
with open(GEOJSON_PATH, "r", encoding="utf-8") as f:
    gj = json.load(f)
# Quick peek (optional)
n_features = len(gj.get("features", []))
print(f"Loaded {n_features} pincode polygons")

literacy = {
    "Andhra Pradesh": 67.0, "Arunachal Pradesh": 65.4, "Assam": 72.2, "Bihar": 61.8,
    "Chhattisgarh": 70.3, "Gujarat": 78.0, "Haryana": 75.6, "Himachal Pradesh": 82.8,
    "Jammu & Kashmir": 67.2, "Jharkhand": 66.4, "Karnataka": 75.4, "Kerala": 94.0,
    "Madhya Pradesh": 69.3, "Maharashtra": 82.3, "Manipur": 76.9, "Meghalaya": 74.4,
    "Mizoram": 91.3, "Nagaland": 79.6, "Odisha": 72.9, "Punjab": 75.8,
    "Rajasthan": 66.1, "Sikkim": 81.4, "Tamil Nadu": 80.1, "Tripura": 87.2,
    "Uttar Pradesh": 67.7, "Uttarakhand": 78.8, "West Bengal": 76.3,
    # UTs
    "Andaman & Nicobar Islands": 86.6, "Chandigarh": 86.0, "Dadra & Nagar Haveli": 76.2,
    "Daman & Diu": 87.1, "Delhi": 86.2, "Goa": 88.7, "Lakshadweep": 91.8, "Puducherry": 85.8,
    # Post-2019 note (for completeness)
    "Ladakh": 67.2  # same as J&K in 2011 vintage
}

# Normalize pincode "Circle" names to state/UT names
# The GeoJSON uses India Post "Circle" (often state-like). We map common variants to census names.
alias_map = {
    "Tamilnadu": "Tamil Nadu",
    "Orissa": "Odisha",
    "Uttaranchal": "Uttarakhand",
    "NCT of Delhi": "Delhi",
    "Pondicherry": "Puducherry",
    "Jammu and Kashmir": "Jammu & Kashmir",
    "Jammu & Kashmir": "Jammu & Kashmir",
    "Andaman and Nicobar Islands": "Andaman & Nicobar Islands",
    "Dadra and Nagar Haveli": "Dadra & Nagar Haveli",
    "Daman and Diu": "Daman & Diu",
    "A & N Islands": "Andaman & Nicobar Islands",
    "A&N Islands": "Andaman & Nicobar Islands",
    "D&NH": "Dadra & Nagar Haveli",
    "Daman & Diu": "Daman & Diu",
}

def canon_state(s):
    if not isinstance(s, str):
        return None
    s = s.strip()
    return alias_map.get(s, s)

# Add a normalized state key to each feature based on its Circle
for feat in gj.get("features", []):
    props = feat.setdefault("properties", {})
    circle = props.get("Circle", "")
    props["state_key"] = canon_state(circle)

#  Build a small DataFrame for Folium join
df = pd.DataFrame({
    "state_key": list(literacy.keys()),
    "literacy": [literacy[k] for k in literacy.keys()]
})

#  Make the map
m = folium.Map(location=[22.8, 79.6], tiles="CartoDB positron", zoom_start=5, control_scale=True)

# Choropleth: color every pincode polygon by its state's literacy rate
choropleth = folium.Choropleth(
    geo_data=gj,
    data=df,
    columns=["state_key", "literacy"],
    key_on="feature.properties.state_key",
    fill_color="YlGn",
    fill_opacity=0.9,
    line_opacity=0.2,
    nan_fill_color="lightgray",
    legend_name="Literacy rate (%) — Census 2011",
    highlight=True
).add_to(m)

# Attach tooltip directly to the choropleth's GeoJson
GeoJsonTooltip(
    fields=["Pincode", "Office_Name", "state_key"],
    aliases=["PIN:", "PO:", "State/UT:"],
    sticky=False,
    labels=True
).add_to(choropleth.geojson)

folium.LayerControl(collapsed=False).add_to(m)
m
````
